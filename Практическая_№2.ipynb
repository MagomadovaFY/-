```python
import math

# функция
def f(x):
    return x**2 + math.sqrt(3) - 3

# метод дихотомии
def dihotomia(a, b, eps=0.01):
    # проверяем знаки на концах
    if f(a) * f(b) > 0:
        raise ValueError("на концах отрезка функция одного знака")
    # проверяем точность
    if eps <= 0:
        raise ValueError("точность должна быть больше нуля")

    n = 0
    while abs(b - a) > eps:
        x = (a + b) / 2
        if f(a) * f(x) < 0:
            b = x
        else:
            a = x
        n += 1
    return (a + b) / 2, n

root, n = dihotomia(1, 1.5)
print(f"метод дихотомии: x = {root:.5f}, итераций: {n}")
```

```python
import math

# функция
def f(x):
    return x**2 + math.sqrt(3) - 3

# производная
def df(x):
    return 2*x

# метод ньютона
def newton(x0, eps=0.01):
    if eps <= 0:
        raise ValueError("точность должна быть больше нуля")
    if df(x0) == 0:
        raise ValueError("производная равна нулю")

    x = x0
    n = 0
    for i in range(100):
        x_new = x - f(x) / df(x)
        n += 1
        if abs(x_new - x) < eps:
            return x_new, n
        x = x_new
    raise ValueError("метод не сошёлся")

root, n = newton(1.5)
print(f"метод ньютона: x = {root:.5f}, итераций: {n}")
```

```python
import math

# функция
def f(x):
    return x**2 + math.sqrt(3) - 3

# производная
def df(x):
    return 2*x

# метод простой итерации
def iteraciya(x0, eps=0.01):
    if eps <= 0:
        raise ValueError("точность должна быть больше нуля")

    # подбираем параметр M
    M = 1.01 * df(x0)
    if M == 0:
        raise ValueError("M равно нулю")

    x = x0
    n = 0
    for i in range(1000):
        x_new = x - f(x) / M
        n += 1
        if abs(x_new - x) < eps:
            return x_new, n
        x = x_new
    raise ValueError("метод не сошёлся")

root, n = iteraciya(1.5)
print(f"метод итерации: x = {root:.5f}, итераций: {n}")
```

```python
# подключаем библиотеки
import math
import matplotlib.pyplot as plt
import numpy as np

# наша функция
def f(x):
    return x**2 + math.sqrt(3) - 3

# производная функции
def df(x):
    return 2*x

# метод дихотомии
def dihotomia(a, b, eps=0.01):
    if f(a) * f(b) > 0:
        raise ValueError("на концах отрезка функция одного знака")
    if eps <= 0:
        raise ValueError("точность должна быть больше нуля")

    n = 0
    while abs(b - a) > eps:
        x = (a + b) / 2
        if f(a) * f(x) < 0:
            b = x
        else:
            a = x
        n += 1
    return (a + b) / 2, n

# метод ньютона
def newton(x0, eps=0.01):
    if eps <= 0:
        raise ValueError("точность должна быть больше нуля")
    if df(x0) == 0:
        raise ValueError("производная равна нулю")

    x = x0
    n = 0
    for i in range(100):
        x_new = x - f(x) / df(x)
        n += 1
        if abs(x_new - x) < eps:
            return x_new, n
        x = x_new
    raise ValueError("метод не сошёлся")

# метод простой итерации
def iteraciya(x0, eps=0.01):
    if eps <= 0:
        raise ValueError("точность должна быть больше нуля")

    M = 1.01 * df(x0)
    if M == 0:
        raise ValueError("M равно нулю")

    x = x0
    n = 0
    for i in range(1000):
        x_new = x - f(x) / M
        n += 1
        if abs(x_new - x) < eps:
            return x_new, n
        x = x_new
    raise ValueError("метод не сошёлся")

# тесты
def test():
    try:
        r1, _ = dihotomia(1, 1.5)
        assert abs(f(r1)) < 0.1, "дихотомия: корень не найден"

        r2, _ = newton(1.5)
        assert abs(f(r2)) < 0.1, "ньютон: корень не найден"

        r3, _ = iteraciya(1.5)
        assert abs(f(r3)) < 0.1, "итерация: корень не найден"

        # проверка на отрицательную точность
        try:
            dihotomia(1, 1.5, eps=-1)
            assert False, "должна быть ошибка при eps < 0"
        except ValueError:
            pass

        # проверка на одинаковые знаки (на отрезке [1.5; 2] корня нет)
        try:
            dihotomia(1.5, 2)
            assert False, "должна быть ошибка при одинаковых знаках"
        except ValueError:
            pass

        print("тесты пройдены успешно")
    except AssertionError as e:
        print(f"ошибка теста: {e}")

# запуск тестов
test()

# считаем всеми методами
r1, n1 = dihotomia(1, 1.5)
r2, n2 = newton(1.5)
r3, n3 = iteraciya(1.5)

print(f"\nметод дихотомии: x = {r1:.5f}, итераций: {n1}")
print(f"метод ньютона:   x = {r2:.5f}, итераций: {n2}")
print(f"метод итерации:  x = {r3:.5f}, итераций: {n3}")

# строим график
x_vals = np.linspace(-2, 2, 500)
y_vals = [f(x) for x in x_vals]

plt.figure(figsize=(10, 5))
plt.plot(x_vals, y_vals, label='f(x) = x^2 + sqrt(3) - 3', color='blue')
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)
plt.scatter([r1], [0], color='red', s=100, zorder=5, label=f'корень = {r1:.4f}')
plt.title('график функции f(x) = x^2 + sqrt(3) - 3')
plt.xlabel('переменная X')
plt.ylabel('переменная Y')
plt.legend()
plt.grid()
plt.show()
```
